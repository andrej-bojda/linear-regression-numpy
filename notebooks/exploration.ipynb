```
import numpy as np
import matplotlib.pyplot as plt
from src.data import generate_data
from src.model import predict, compute_loss, compute_gradients
from src.train import train
X, y, w_true, b_true = generate_data(w_true = 2.5, b_true = 1.0)
epochs = 450
w, b, loss_history = train(X = X, y = y, alpha = 0.1, epochs = epochs)
```

```
plt.plot(range(epochs) , loss_history, color = "blue", linewidth = 2)
plt.title("Model training progress")
plt.xlabel("Epochs(Time)")
plt.ylabel("Loss(Error)")
plt.savefig("../plots/loss_curve.png", dpi = 150, bbox_inches = "tight")
```

```
plt.scatter(X, y, color = "blue")
plt.plot(X, predict(X, w, b), color = "red")
plt.title("Linear regression results")
plt.xlabel("Independent variable (X)")
plt.ylabel("Target variable (Y)")
plt.savefig("../plots/fit.png", dpi = 150, bbox_inches = "tight")
```

```
epochs = 450
alphas = [0.01, 0.1, 0.5]
color = ["blue", "red", "green"]
for alpha, col in zip(alphas, color):
    w, b, loss_history = train(X, y, alpha = alpha, epochs = epochs)
    plt.plot(range(epochs), loss_history, color = col, linewidth = 2, label=f"Alpha {alpha}")
plt.title("Model training progress with different learning curves")
plt.xlabel("Epochs(Time)")
plt.ylabel("Loss(Error)")
plt.legend(labelcolor='linecolor')
plt.savefig("../plots/lr_sensitivity.png", dpi = 150, bbox_inches = "tight")
```
